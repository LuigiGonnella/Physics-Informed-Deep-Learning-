### ADD 3D AUGMENTATION

The 3D rotation matrices about the x, y, and z axes are:

```math
R_x(\theta)=
\begin{pmatrix}
1 & 0 & 0 \\
0 & \cos\theta & -\sin\theta \\
0 & \sin\theta & \cos\theta
\end{pmatrix},
```

```math
R_y(\theta)=
\begin{pmatrix}
\cos\theta & 0 & \sin\theta \\
0 & 1 & 0 \\
-\sin\theta & 0 & \cos\theta
\end{pmatrix},
```

```math
R_z(\theta)
=
\begin{pmatrix}
\cos\theta & -\sin\theta & 0 \\
\sin\theta & \cos\theta & 0 \\
0 & 0 & 1
\end{pmatrix}.
```

## ANALYSIS

We achieve better results, but still **far way from SOTA** performance --> data augmentation helps but **does not make MLP rotationally invariant**, it just helps to generalize on it. It does not solve all other symmetries yet, for this, we'd need a **GNN**.

Alternatively, if we want to maintain the _same architecture_, we could change only the **INPUT REPRESENTATION** to something **rotationally invariant**, e.g. instead of R we could use **pairwise distances** between each atom (i row on N rows) in a single molecula ((N, 3))

```math

d_{ij} = \|\mathbf r_i-\mathbf r_j\|_2.

```

these are INVARIANT to rotations, in fact **d(r) = d(Qr)** --> model sees only D, which is the same for r and Qr --> augmentation not needed (useless), and so:

```math
E(Qd(R))=E(d(R)).
```

- if instead of a _scalar_ `E_target` we have a _VECTOR FORCE_ `F`, we'd desire **EQUIVARIANCE**:
```math
\mathbf F_i(QR)
=
Q\,\mathbf F_i(R).
```
and this is not achievable with an MLP, swithc to **GNN** instead.

So, **recap**:
if we apply the same force on a rotated molecula we want the same result of rotating the result of the force on the non-rotated molecule
--> If we rotate the molecule by Q, the **physical force should rotate in the same way**
--> If we rotate the molecule by Q, the **energy should remain the same**
