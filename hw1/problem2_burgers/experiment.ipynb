````python
dataset = BurgersDataset(data_path = 'data/Burgers_train_1000_visc_0.01.mat')

n_samples = len(dataset)
n_samples
````

````python
import matplotlib.pyplot as plt
from matplotlib.figure import Figure

def plot_sol(ax, sample : torch.tensor, sol : torch.tensor) -> Figure:

    x = sample[:, :, 0].detach().cpu().numpy() #x values
    t = sample[:, :, 1].detach().cpu().numpy() #t values
    u = sol.detach().cpu().numpy()
    
    ax.set_xlabel("Time t")
    ax.set_ylabel("Position x")
    mesh = ax.pcolormesh(t, x, u, shading = "nearest", cmap = "viridis")
    
    return mesh
````

````python
show_samples = 5

for i in range(show_samples):
    sample, sol = dataset[i]

    fig, ax = plt.subplots()

    mesh = plot_sol(ax, sample, sol)
    fig.colorbar(mesh, ax=ax)

````

- Different initial profiles: some have a positive peak near x ~ 0.2; others are positive near the ends of the spatial domain and negative near the middle.
- The colors become less extreme over time: peaks weaken, troughs become shallower, and the profiles smooth out. That is consistent with viscosity in Burgers’ equation.
- The bands bend as time increases: features change position as well as amplitude, consistent with nonlinear transport.
- Some transitions stay fairly steep, especially around x ~ 0.2 in images 1, 4, and 5, though they remain smooth rather than becoming discontinuous.

## ADD PDE LOSS 

### PRED LOSS ON u_real

````python
from pde import burgers_pde_loss

train_dataset = BurgersDataset('data/Burgers_train_1000_visc_0.01.mat', train=False)

train_dataloader = DataLoader(train_dataset, batch_size=len(train_dataset))


batch = next(iter(train_dataloader))
samples, u = batch

x = samples[:, :, :, 0]
t = samples[:, :, :, 1]

pde_loss = burgers_pde_loss(x, t, u, per_batch = True) 

print(f"TRUE u PDE LOSS STD: {torch.std(pde_loss).item()}")
print(f"TRUE u PDE LOSS MEAN {torch.mean(pde_loss).item()}")

````

The `PDE residual loss` is expected to be `small` but _not exactly zero_. Although the dataset contains solutions of the Burgers equation, the `derivatives` used to compute the residual are `approximated` using `finite differences`, which introduce truncation error. In addition, the reference solutions were generated numerically and may use a different discretization scheme, introducing further numerical error.

### PDE LOSS ON U_pred

````python

batch = next(iter(train_dataloader))
samples = batch[0]

model.eval()
with torch.no_grad():
    u_pred = model(samples)

x = samples[:, :, :, 0]
t = samples[:, :, :, 1]

pde_loss = burgers_pde_loss(x, t, u_pred, per_batch = True) 

print(f"PREDICTED u PDE LOSS STD: {torch.std(pde_loss).item()}")
print(f"PREDICTED u PDE LOSS MEAN {torch.mean(pde_loss).item()}")

````

### TRAIN WITH ONLY PDE_LOSS

````python
model_pde, train_losses_pde, val_losses_pde = train(loss_type = 'pde')
````

````python
fig, ax = plt.subplots()

train_losses_pde = np.array(train_losses_pde)
val_losses_pde = np.array(val_losses_pde)

t = np.arange(train_losses_pde.shape[0])

ax.plot(t, train_losses_pde, color = 'red', label = 'train_loss_pde')
ax.plot(t, val_losses_pde, color = 'blue', label = 'val_losses_pde')

ax.set_xlabel("steps")
ax.set_ylabel("loss")

ax.legend()
ax.grid()
fig.tight_layout()
````

````python
test_dataset = BurgersDataset(
    "data/Burgers_test_50_visc_0.01.mat",
    train=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=len(test_dataset)
)

samples, solutions = next(iter(test_loader))

samples = samples
solutions = solutions

model_pde.eval()

with torch.no_grad():
    u_pred = model_pde(samples)

    x = samples[:, :, :, 0]
    t = samples[:, :, :, 1]

    # one data loss for each test sample
    err = u_pred - solutions
    err_norm = torch.linalg.vector_norm(err, dim = (1, 2))
    target_norm = torch.linalg.vector_norm(solutions, dim = (1, 2))

    data_losses = err_norm / target_norm.clamp_min(1e-12)

    # one PDE loss for each test sample
    pde_losses = burgers_pde_loss(
        x, t, u_pred,
        per_batch=True
    )

print("DATA LOSS")
print("mean:", data_losses.mean().item())
print("std :", data_losses.std().item())

print("PDE LOSS")
print("mean:", pde_losses.mean().item())
print("std :", pde_losses.std().item())
````

Training only on PDE gives us HIGH DATA LOSS (121 % rel. err.) and LOW PDE LOSS. So in either case a single loss is not enough --> we need both! 
Anyway adding the INITIAL CONDITIONS and the BOUNDARY CONDITION to our PDE LOSS would improve the result.

e.g. 
```python
u0 = solution[:, :, 0]
ic_loss = torch.mean(
    (u_pred[:, :, 0] - u0) ** 2
)
```

and

```python
bc_value_loss = torch.mean(
    (u_pred[:, 0, :] - u_pred[:, -1, :]) ** 2
)
```

## TUNE DIFFERENT ALPHAS

````python
alphas = [0.1, 0.2, 0.5, 0.7]
models = []

for alpha in alphas:
    model, _, _ = train(loss_type = 'both', alpha = alpha)
    models.append(model)
````

````python
samples, solutions = next(iter(test_loader))

for i, model in enumerate(models):

    model.eval()

    with torch.no_grad():
        u_pred = model(samples)

        x = samples[:, :, :, 0]
        t = samples[:, :, :, 1]

        # one data loss for each test sample
        err = u_pred - solutions
        err_norm = torch.linalg.vector_norm(err, dim = (1, 2))
        target_norm = torch.linalg.vector_norm(solutions, dim = (1, 2))

        data_losses = err_norm / target_norm.clamp_min(1e-12)

        # one PDE loss for each test sample
        pde_losses = burgers_pde_loss(
            x, t, u_pred,
            per_batch=True
        )

    print(f"alpha: {alphas[i]} - DATA LOSS:")
    print("mean:", data_losses.mean().item())
    print("std :", data_losses.std().item())

    print(f"alpha: {alphas[i]} - PDE LOSS")
    print("mean:", pde_losses.mean().item())
    print("std :", pde_losses.std().item())
````
